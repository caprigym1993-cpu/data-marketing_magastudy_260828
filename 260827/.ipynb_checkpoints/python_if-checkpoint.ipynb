1. 조건문
- if 단독사용 조건식 (t/f) 실행문
- 논리연산자3 종류 
- a and d 에이조건도 참 비조건도 참  에이 비 모두 참이어야만 조건식 참 
- a or d 에이 조건 혹은 비 조건 중 하나라도 참이면 참 = 에이 비 둘중 하나만 참이면 조건식 참 
- not a  에이가 거짓이면 전체 조건식은 참 ㅣ 에이가 참이면 전체 조건식 거짓  

```python
age = 20

if age >= 19 :
    print(f"{age}세당신은 성인 ")
```

```python
age3 = 20

if age >= 19:
    print("당신은 성인입니다 ")
else:
    print("청소년")
```

```python
age4 = 17

if age4>= 19 :
    print("성인")
elif age4 < 19 and age4 >= 13 : 
    print("청소년")
else : 
    print("유소년")
```

```python
# 중첩 이프 조건문: 이프 조건문 > 또 다른 이프 조건문 존재
cash = 120000 
if cash > 100000 : 
    print ("gg")
else : 
    if cash > 50000 :
        print ("쇼핑")
    else : 
        print("집가")
```

```python
# 엘이프 : 복수의 조건을 모두 동시에 따져보고자 할때 
# 변수 a에 값을 먼저 할당
a = 19  

if a > 20:
    print("에이스")
elif a > 18:
    print("굿")
elif a > 17:
    print("별로")
else:
    print("nn")

#메모리 누수 개념 
```
