```python
#타입 :함수
#카운트 : 메서드
#함수는 앞에 어떤 것의 도움을 받지 않고 독립적으로 사용 가능
#메서드는 사전적의미 방법  , 거의 대부분 앞에 어떤것 (=객체) 의 도움을 받아야함 

print(article . count("Korea")) # 인자값으로 이력된 문자열이 몇 번 사용되었는가(대 소문자 구분함  )
print(len(article))# 전체 문자열 길이 갯수 (빈칸 ,트수문자 모두 포함 
print(article.find("S"))# 전체 문자열 중 몇 번째 인덱스 = 순번 값으로 해당 인자값이 사용되었는지 확인 
```

```python
index = article.find("s")

while index != -1:
    print(index)
    index = article.find("s", index + 1)  # '=' 앞에 index 변수명 추가
```
